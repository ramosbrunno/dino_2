```python
# Teste 3: Extrair token - sua linha exata
print("🔑 Teste 3: Extrair token")
print("=" * 25)

try:
    # SUA LINHA EXATA:
    admin_token = dbutils.notebook.entry_point.getDbutils().notebook().getContext().apiToken().get()
    
    if admin_token:
        # Mascarar para segurança
        masked = f"{admin_token[:15]}...{admin_token[-8:]}" if len(admin_token) > 23 else "***"
        print(f"✅ Token extraído: {masked}")
        print(f"   Comprimento: {len(admin_token)}")
        
        # Verificar formato
        if admin_token.startswith('dapi'):
            print("✅ Formato Databricks correto (dapi...)")
        else:
            print(f"⚠️ Formato inesperado: {admin_token[:10]}...")
        
        # Salvar para próximo teste
        globals()['admin_token'] = admin_token
        print(f"🔑 Token salvo para próximos testes")
    else:
        print("⚠️ Token está vazio")
        
except Exception as e:
    print(f"❌ ERRO: {e}")
    print(f"   Tipo do erro: {type(e).__name__}")
    import traceback
    traceback.print_exc()
```
