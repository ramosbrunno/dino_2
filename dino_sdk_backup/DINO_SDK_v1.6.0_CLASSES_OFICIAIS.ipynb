### ✅ SOLUÇÃO IMPLEMENTADA v1.6.0:

Baseado na **sua referência oficial da documentação**, agora usamos:

```python

# ✅ CORRETO - Classes oficiais
task = Task(
    task_key="my-key",
    notebook_task=NotebookTask(
        notebook_path="/path/to/notebook",
        source=Source("WORKSPACE")
    ),
    new_cluster=NewCluster(
        spark_version="13.3.x-scala2.12",
        node_type_id="Standard_DS3_v2",
        num_workers=1
    )
)

job = client.jobs.create(
    name="job-name",
    tasks=[task]  # Lista de objetos Task, não dicionários!
)
```
