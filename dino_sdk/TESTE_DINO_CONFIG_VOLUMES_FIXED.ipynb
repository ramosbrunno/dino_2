# 🧪 Teste DINO-Config com Volumes

## Volumes Criados Automaticamente

- **_schemas**: Para schemas do AutoLoader

- ✅ Criação de schema com localização gerenciada
- ✅ Criação de volumes gerenciados automaticamente
- ✅ Verificação de existência antes da criação

## 🔌 2. Importar Bibliotecas

```python
from pyspark.sql import SparkSession
from dino_sdk.schema_manager import SchemaManager, create_schema_simple, ensure_schema_simple

print("📚 Bibliotecas importadas com sucesso!")
print(f"🌟 Spark Session: {spark.version}")
```

## 🧪 3. Teste 1 - Criar Schema Bronze com Volumes

```python
print("=" * 60)
print("🧪 TESTE 1: Criando Schema Bronze com Volumes")
print("=" * 60)

# Configuração
catalog_name = "main"
schema_name = "bronze_test_volumes"

print(f"📋 Parâmetros:")
print(f"   • Catálogo: {catalog_name}")
print(f"   • Schema: {schema_name}")
print()

# Método 1: Usando classe SchemaManager
try:
    manager = SchemaManager(catalog_name, schema_name)
    result = manager.create_schema(spark)
    
    print("📊 Resultado da Criação:")
    print(f"   • Sucesso: {result['success']}")
    print(f"   • Schema criado: {result['schema_created']}")
    print(f"   • Já existia: {result['already_exists']}")
    
    if 'volumes_created' in result and result['volumes_created']:
        print(f"   • Volumes criados: {result['volumes_created']}")
    
    if 'volumes_existing' in result and result['volumes_existing']:
        print(f"   • Volumes já existentes: {result['volumes_existing']}")
    
    if result['errors']:
        print(f"   • Erros: {result['errors']}")
    
    print()
    
    if result['success']:
        print("✅ Schema Bronze criado com sucesso!")
    else:
        print("❌ Falha na criação do schema")
        
except Exception as e:
    print(f"❌ Erro durante o teste: {str(e)}")
```

## 🔍 4. Verificar Volumes Criados

```python
print("=" * 60)
print("🔍 VERIFICAÇÃO: Listando Volumes Criados")
print("=" * 60)

# Listar volumes no schema bronze
print(f"📦 Volumes no schema {catalog_name}.{schema_name}:")
try:
    volumes_df = spark.sql(f"SHOW VOLUMES IN {catalog_name}.{schema_name}")
    volumes = volumes_df.collect()
    
    if volumes:
        for volume in volumes:
            print(f"   ✅ {volume.volume_name}")
    else:
        print("   ⚠️ Nenhum volume encontrado")
        
except Exception as e:
    print(f"   ❌ Erro ao listar volumes: {str(e)}")

print()
```

## 🧪 5. Teste 2 - Usar ensure_schema_simple (Recomendado)

```python
print("=" * 60)
print("🧪 TESTE 2: Usando ensure_schema_simple (Função Recomendada)")
print("=" * 60)

# Configuração para um novo schema
catalog_name = "main"
schema_name = "silver_test_volumes"

print(f"📋 Parâmetros:")
print(f"   • Catálogo: {catalog_name}")
print(f"   • Schema: {schema_name}")
print()

# Usando função de conveniência ensure_schema_simple
try:
    result = ensure_schema_simple(spark, catalog_name, schema_name)
    
    print("📊 Resultado da Operação:")
    print(f"   • Sucesso: {result['success']}")
    print(f"   • Schema criado: {result['schema_created']}")
    print(f"   • Já existia: {result['already_exists']}")
    
    if 'volumes_created' in result and result['volumes_created']:
        print(f"   • Volumes criados: {result['volumes_created']}")
    
    if 'volumes_existing' in result and result['volumes_existing']:
        print(f"   • Volumes já existentes: {result['volumes_existing']}")
    
    if result['errors']:
        print(f"   • Erros: {result['errors']}")
    
    if 'volume_errors' in result:
        print(f"   • Erros nos volumes: {result['volume_errors']}")
    
    print()
    
    if result['success']:
        print("✅ Schema Silver configurado com sucesso!")
    else:
        print("❌ Falha na configuração do schema")
        
except Exception as e:
    print(f"❌ Erro durante o teste: {str(e)}")
```

## 🧪 6. Teste 3 - Executar Novamente (Verificar Idempotência)

```python
print("=" * 60)
print("🧪 TESTE 3: Executando Novamente (Teste de Idempotência)")
print("=" * 60)

# Executar novamente no mesmo schema para testar idempotência
catalog_name = "main"
schema_name = "silver_test_volumes"

print(f"📋 Executando novamente para: {catalog_name}.{schema_name}")
print()

try:
    result = ensure_schema_simple(spark, catalog_name, schema_name)
    
    print("📊 Resultado da Segunda Execução:")
    print(f"   • Sucesso: {result['success']}")
    print(f"   • Schema criado: {result['schema_created']}")
    print(f"   • Já existia: {result['already_exists']}")
    
    if 'volumes_created' in result and result['volumes_created']:
        print(f"   • Volumes criados: {result['volumes_created']}")
    
    if 'volumes_existing' in result and result['volumes_existing']:
        print(f"   • Volumes já existentes: {result['volumes_existing']}")
    
    print()
    
    if result['success'] and result['already_exists']:
        print("✅ Idempotência confirmada - Schema e volumes já existiam!")
    elif result['success']:
        print("✅ Operação executada com sucesso!")
    else:
        print("❌ Falha na segunda execução")
        
except Exception as e:
    print(f"❌ Erro durante o teste de idempotência: {str(e)}")
```

## 🧪 7. Teste 4 - Schema com Localização Personalizada

```python
print("=" * 60)
print("🧪 TESTE 4: Schema com Localização Gerenciada Personalizada")
print("=" * 60)

# Configuração com localização personalizada
catalog_name = "main"
schema_name = "gold_test_volumes"

# IMPORTANTE: Ajuste o caminho abaixo para sua configuração
# Este é um exemplo - use o caminho correto do seu environment
managed_location = "/Volumes/main/default/system_files/managed_schemas/gold_test_volumes/"

print(f"📋 Parâmetros:")
print(f"   • Catálogo: {catalog_name}")
print(f"   • Schema: {schema_name}")
print(f"   • Localização: {managed_location}")
print()

try:
    # Usar função com localização personalizada
    result = ensure_schema_simple(spark, catalog_name, schema_name, managed_location)
    
    print("📊 Resultado com Localização Personalizada:")
    print(f"   • Sucesso: {result['success']}")
    print(f"   • Schema criado: {result['schema_created']}")
    print(f"   • Já existia: {result['already_exists']}")
    
    if 'volumes_created' in result and result['volumes_created']:
        print(f"   • Volumes criados: {result['volumes_created']}")
    
    if 'volumes_existing' in result and result['volumes_existing']:
        print(f"   • Volumes já existentes: {result['volumes_existing']}")
    
    if result['errors']:
        print(f"   • Erros: {result['errors']}")
    
    print()
    
    if result['success']:
        print("✅ Schema Gold com localização personalizada criado!")
    else:
        print("❌ Falha na criação com localização personalizada")
        
except Exception as e:
    print(f"❌ Erro durante o teste com localização: {str(e)}")
    print("💡 Dica: Verifique se o caminho da localização está correto")
```

## 📊 8. Relatório Final - Listar Todos os Schemas e Volumes

```python
print("=" * 60)
print("📊 RELATÓRIO FINAL - Schemas e Volumes Criados")
print("=" * 60)

# Listar todos os schemas de teste criados
test_schemas = ["bronze_test_volumes", "silver_test_volumes", "gold_test_volumes"]
catalog_name = "main"

for schema_name in test_schemas:
    print(f"\n🗂️ Schema: {catalog_name}.{schema_name}")
    
    # Verificar se schema existe
    try:
        schema_check = spark.sql(f"SHOW SCHEMAS IN {catalog_name} LIKE '{schema_name}'").count()
        
        if schema_check > 0:
            print(f"   ✅ Schema existe")
            
            # Listar volumes no schema
            try:
                volumes_df = spark.sql(f"SHOW VOLUMES IN {catalog_name}.{schema_name}")
                volumes = volumes_df.collect()
                
                if volumes:
                    print(f"   📦 Volumes encontrados:")
                    for volume in volumes:
                        print(f"      • {volume.volume_name}")
                else:
                    print(f"   ⚠️ Nenhum volume encontrado")
                    
            except Exception as e:
                print(f"   ❌ Erro ao listar volumes: {str(e)}")
        else:
            print(f"   ⚠️ Schema não existe")
            
    except Exception as e:
        print(f"   ❌ Erro ao verificar schema: {str(e)}")

print("\n" + "=" * 60)
print("🎉 TESTE DINO-CONFIG COM VOLUMES CONCLUÍDO!")
print("=" * 60)

print("\n💡 Funcionalidades testadas:")
print("   ✅ Criação de schemas com volumes automáticos")
print("   ✅ Volumes gerenciados (_checkpoints, _schemas, raw)")
print("   ✅ Verificação de existência (idempotência)")
print("   ✅ Schema com localização gerenciada personalizada")
print("   ✅ Tratamento de erros e logs detalhados")

print("\n🚀 O DINO SDK está pronto para uso com gerenciamento de volumes!")
```
