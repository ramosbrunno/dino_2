## 🔍 **Teste 1: Verificar Ambiente Databricks**

````python
# ✅ Teste 1: Verificar se dbutils está disponível no ambiente
print("🔍 TESTE 1: Verificando ambiente Databricks...")
print("-" * 50)

try:
    # Verificar se dbutils existe
    dbutils_available = 'dbutils' in globals()
    print(f"📊 dbutils em globals(): {dbutils_available}")
    
    if dbutils_available:
        print(f"✅ dbutils encontrado: {type(dbutils)}")
        
        # Extrair informações básicas
        try:
            workspace_url = dbutils.notebook.entry_point.getDbutils().notebook().getContext().apiUrl().get()
            print(f"🌐 Workspace URL: https://{workspace_url}")
            
            token = dbutils.notebook.entry_point.getDbutils().notebook().getContext().apiToken().get()
            print(f"🔑 Token disponível: {str(token)[:20]}...{str(token)[-10:]}")
            
            print("✅ TESTE 1 PASSOU: Ambiente Databricks válido")
            
        except Exception as e:
            print(f"⚠️ Erro ao extrair credenciais: {e}")
            print("⚠️ TESTE 1 PARCIAL: dbutils existe mas credenciais falharam")
    else:
        print("❌ dbutils não encontrado em globals()")
        print("❌ TESTE 1 FALHOU: Não é ambiente Databricks válido")
        
except Exception as e:
    print(f"❌ Erro no teste: {e}")
    print("❌ TESTE 1 FALHOU: Erro inesperado")

print("\n" + "="*60)
````

## 🤖 **Teste 2: Testar Auto-detecção do SDK v1.1.2**

````python
# ✅ Teste 2: Testar auto-detecção robusta do SDK v1.1.3
print("🤖 TESTE 2: Auto-detecção robusta SDK v1.1.3...")
print("🔍 FOCO: Encontrar apenas dbutils NATIVO (não o do SDK)")
print("-" * 50)

try:
    # Importar função de detecção
    from src.keyvault_config import get_notebook_context
    print("✅ Módulo importado com sucesso")
    
    # Testar detecção automática robusta
    print("🔍 Executando get_notebook_context() com detecção robusta...")
    context = get_notebook_context()
    
    if context and 'workspace_url' in context and 'token' in context:
        print(f"✅ Workspace URL: {context['workspace_url']}")
        print(f"✅ Token extraído: {str(context['token'])[:20]}...{str(context['token'])[-10:]}") 
        print(f"✅ dbutils objeto: {type(context.get('dbutils', 'N/A'))}")
        
        # Verificar se é dbutils nativo
        dbutils_obj = context.get('dbutils')
        if dbutils_obj:
            dbutils_type = str(type(dbutils_obj))
            if "module" in dbutils_type.lower():
                print("⚠️ ATENÇÃO: dbutils retornado é um módulo (SDK)")
            else:
                print("✅ SUCESSO: dbutils retornado é um objeto (nativo)")
                
        print("🎉 TESTE 2 PASSOU: Auto-detecção robusta funcionou!")
        
        # Salvar contexto para próximos testes
        detected_context = context
        auto_detection_success = True
        
    elif context:
        print(f"⚠️ Contexto parcial retornado: {list(context.keys())}")
        print("⚠️ TESTE 2 PARCIAL: Auto-detecção incompleta")
        auto_detection_success = False
        
    else:
        print(f"❌ Contexto retornado: {context}")
        print("❌ TESTE 2 FALHOU: Auto-detecção não funcionou")
        auto_detection_success = False
        
except Exception as e:
    print(f"❌ Erro na auto-detecção: {e}")
    import traceback
    print("📊 Traceback:")
    traceback.print_exc()
    print("❌ TESTE 2 FALHOU: Erro na importação/execução")
    auto_detection_success = False

print("\n" + "="*60)
````

````python
# ✅ Teste 1.5: Verificar tipos de dbutils disponíveis
print("🔍 TESTE 1.5: Analisando tipos de dbutils disponíveis...")
print("-" * 50)

try:
    # Verificar diferentes tipos de dbutils
    dbutils_types = []
    
    # Verificar em globals
    if 'dbutils' in globals():
        dbutils_global = globals()['dbutils']
        dbutils_types.append(f"globals(): {type(dbutils_global)}")
        print(f"📊 dbutils em globals(): {type(dbutils_global)}")
        
        # Verificar se é módulo (SDK) ou objeto (nativo)
        if str(type(dbutils_global)).lower().find('module') >= 0:
            print("⚠️ dbutils em globals() é um MÓDULO (SDK)")
        else:
            print("✅ dbutils em globals() é um OBJETO (possivelmente nativo)")
            
        # Verificar atributos
        if hasattr(dbutils_global, 'notebook'):
            print(f"✅ Tem atributo 'notebook': {type(dbutils_global.notebook)}")
            if hasattr(dbutils_global.notebook, 'entry_point'):
                print("✅ Tem 'notebook.entry_point' (INTERFACE NATIVA)")
            else:
                print("⚠️ Não tem 'notebook.entry_point'")
        else:
            print("❌ Não tem atributo 'notebook'")
    
    # Tentar eval
    try:
        dbutils_eval = eval('dbutils')
        dbutils_types.append(f"eval(): {type(dbutils_eval)}")
        print(f"📊 dbutils via eval(): {type(dbutils_eval)}")
        
        if str(type(dbutils_eval)).lower().find('module') >= 0:
            print("⚠️ dbutils via eval() é um MÓDULO (SDK)")
        else:
            print("✅ dbutils via eval() é um OBJETO (possivelmente nativo)")
    except Exception as e:
        print(f"❌ eval('dbutils') falhou: {e}")
    
    # Tentar importar do SDK
    try:
        from databricks.sdk import dbutils as sdk_dbutils
        dbutils_types.append(f"SDK import: {type(sdk_dbutils)}")
        print(f"📊 dbutils do SDK: {type(sdk_dbutils)}")
        print("⚠️ Este é o dbutils do SDK (módulo)")
    except Exception as e:
        print(f"ℹ️ SDK dbutils não importado: {e}")
    
    print(f"\n🔍 Resumo de tipos encontrados:")
    for dtype in dbutils_types:
        print(f"   - {dtype}")
        
    print("✅ TESTE 1.5 CONCLUÍDO: Análise de tipos")
    
except Exception as e:
    print(f"❌ Erro no teste de tipos: {e}")

print("\n" + "="*60)
````

## 💉 **Teste 3: Injeção Manual (Fallback)**

````python
# ✅ Teste 3: Testar injeção manual
print("💉 TESTE 3: Injeção Manual de Contexto...")
print("-" * 50)

try:
    # Se auto-detecção falhou, extrair credenciais manualmente
    if not auto_detection_success:
        print("⚠️ Auto-detecção falhou, extraindo credenciais manualmente...")
        
        # Extrair diretamente do dbutils
        if 'dbutils' in globals():
            workspace_instance = dbutils.notebook.entry_point.getDbutils().notebook().getContext().apiUrl().get()
            admin_token = dbutils.notebook.entry_point.getDbutils().notebook().getContext().apiToken().get()
            workspace_url = f"https://{workspace_instance}"
            
            print(f"✅ URL extraída: {workspace_url}")
            print(f"✅ Token extraído: {str(admin_token)[:20]}...{str(admin_token)[-10:]}")
            
            # Importar função de injeção
            from src.keyvault_config import inject_notebook_context
            
            # Injetar contexto
            inject_notebook_context(workspace_url, admin_token, dbutils)
            print("✅ Contexto injetado no SDK")
            
            # Testar se injeção funcionou
            from src.keyvault_config import get_notebook_context
            injected_context = get_notebook_context()
            
            if injected_context and 'workspace_url' in injected_context:
                print(f"✅ Contexto injetado verificado: {injected_context['workspace_url']}")
                print("🎉 TESTE 3 PASSOU: Injeção manual funcionou!")
                manual_injection_success = True
                final_context = injected_context
            else:
                print("❌ Injeção não funcionou")
                manual_injection_success = False
        else:
            print("❌ dbutils não disponível para extração manual")
            manual_injection_success = False
    else:
        print("✅ Auto-detecção funcionou, injeção manual não necessária")
        manual_injection_success = True
        final_context = detected_context
        
except Exception as e:
    print(f"❌ Erro na injeção manual: {e}")
    print("❌ TESTE 3 FALHOU: Erro na injeção")
    manual_injection_success = False

print("\n" + "="*60)
````

## 🔧 **Teste 4: Inicializar KeyVaultConfigManager**

````python
# ✅ Teste 4: Testar KeyVaultConfigManager
print("🔧 TESTE 4: Inicializando KeyVaultConfigManager...")
print("-" * 50)

try:
    # Verificar se temos contexto válido
    if auto_detection_success or manual_injection_success:
        print("✅ Contexto válido disponível, inicializando SDK...")
        
        # Importar e inicializar KeyVaultConfigManager
        from src.keyvault_config import KeyVaultConfigManager
        
        kv_manager = KeyVaultConfigManager(
            keyvault_name="dino-shared-keyvault",
            catalog_name="dino_catalog",
            schema_name="teste_v112"
        )
        
        print("✅ KeyVaultConfigManager criado com sucesso!")
        
        # Verificar se cliente Databricks foi inicializado
        if hasattr(kv_manager, 'databricks_client') and kv_manager.databricks_client:
            print("🎉 CLIENTE DATABRICKS INICIALIZADO!")
            
            # Tentar obter informações do usuário
            try:
                user_info = kv_manager.databricks_client.current_user.me()
                print(f"👤 Usuário: {user_info.user_name}")
                print(f"📧 Email: {user_info.user_name}")
            except Exception as e:
                print(f"⚠️ Não foi possível obter info do usuário: {e}")
            
            # Tentar listar secret scopes
            try:
                scopes = kv_manager.databricks_client.secrets.list_scopes()
                scope_names = [scope.name for scope in scopes]
                print(f"📋 Secret Scopes: {len(scope_names)} acessíveis")
                if scope_names:
                    print(f"📌 Scopes: {scope_names[:3]}...")  # Primeiros 3
            except Exception as e:
                print(f"⚠️ Não foi possível listar scopes: {e}")
            
            print("🎉 TESTE 4 PASSOU: SDK totalmente funcional!")
            sdk_success = True
            
        else:
            print("❌ Cliente Databricks não foi inicializado")
            print("❌ TESTE 4 FALHOU: SDK não funcional")
            sdk_success = False
            
    else:
        print("❌ Não há contexto válido disponível")
        print("❌ TESTE 4 CANCELADO: Sem contexto")
        sdk_success = False
        
except Exception as e:
    print(f"❌ Erro ao inicializar SDK: {e}")
    print("❌ TESTE 4 FALHOU: Erro na inicialização")
    sdk_success = False

print("\n" + "="*60)
````
